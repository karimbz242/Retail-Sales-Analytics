```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_quantity = df['Quantity'].sum()
total_orders = df['Order ID'].nunique()

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Quantity Sold: {total_quantity:,}")
print(f"Total Orders: {total_orders:,}")
```

```python
df['Order ID'].nunique()
```

```python
average_order_value = total_sales / total_orders
profit_margin = (total_profit / total_sales) * 100

print(f"Average Order Value: ${average_order_value:,.2f}")
print(f"Profit Margin: {profit_margin:.2f}%")
```

## 14. Final Business Insights

### Key Findings

- Total sales reached approximately **$2.30 million**, with total profit of approximately **$286,397**.
- **2018** was the strongest year for sales and profit.
- **Technology** generated the highest sales and profit among the three major categories.
- **Copiers, Phones, and Accessories** were the strongest profit-generating sub-categories.
- **Tables** generated the largest loss among sub-categories.
- Discounts of **30% or higher were associated with negative overall profit margins**.
- The **West** had the highest regional profit margin, while the **Central** region had the lowest.
- **November and December** were the strongest sales months, while December generated the highest monthly profit.
- The business should evaluate customers using both **sales and profitability**, since high sales do not always result in high profit.

## 15. Business Recommendations

### 1. Review High Discounts

Discounts of 30% or more were consistently associated with negative profitability. Management should review high-discount transactions and evaluate whether discounts can be reduced or targeted more selectively.

### 2. Investigate Loss-Making Products

Tables and Bookcases generated negative profit. Product pricing, costs, discounts, and demand should be reviewed to determine why these products are unprofitable.

### 3. Focus on High-Profit Products

Copiers, Phones, and Accessories generated substantial profit. These products should be evaluated for additional promotional and sales opportunities.

### 4. Investigate Regional Performance

The Central region had the lowest profit margin. Management should investigate pricing, discounts, shipping costs, and product mix in this region.

### 5. Prepare for Seasonal Demand

Sales increased significantly from September through December. Inventory, staffing, and marketing resources should be planned ahead of this period.

### 6. Monitor Customer Profitability

Customers should be evaluated using both revenue and profit. A customer generating high sales may not necessarily be highly profitable.
